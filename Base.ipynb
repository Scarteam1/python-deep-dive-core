```python
print(int("10", base=10)) # Base 10: 1 group of ten + 0 loose ones  -> 10
print(int("10", base=2))  # Base 2 : 1 group of two  + 0 loose ones  -> 2
print(int("10", base=8))  # Base 8 : 1 group of eight + 0 loose ones -> 8
print(int("10", base=16)) # Base 16: 1 group of sixteen + 0 loose ones -> 16
```
