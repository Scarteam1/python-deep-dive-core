```python
class Rectangle:
    def __init__(self, Length, Width):
        self.Length = Length
        self.Width = Width

    def area(self):
        return self.Length * self.Width

    def perimeter(self):
        return 2 * (self.Length + self.Width)

    def __str__(self):
        return 'This is a string'.format (self.Length, self.Width)
```

```python
class Rectangle:
    def __init__(self, Length, Width):
        self.Length = Length
        self.Width = Width

    def area(self):
        return self.Length * self.Width

    def perimeter(self):
        return 2 * (self.Length + self.Width)

    def __str__(self):
        return 'This is a string'.format (self.Length, self.Width)

    def __eq__(self, other):
        return self.Length == other.Length and self.Width == other.Width
```

```python
r1 = Rectangle(10,20)
r2 = Rectangle(10,20)
```

```python
r1 is not r2
```

```python
r1 == r2
```
