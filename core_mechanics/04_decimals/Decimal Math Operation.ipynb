```python
from decimal import Decimal

x = 2
y = 0.5

Output = x * y
print(Output)
```

```python
from decimal import Decimal, getcontext

context = getcontext()
context.prec = 4

x = 2
y = Decimal(0.05)

Output = x * y
print(Output)
```

```python
from decimal import Decimal, getcontext

context = getcontext()
context.prec = 4

x = 2
y = 0.05
z = Decimal(10.05)

Output = z * y
print(Output)
```
