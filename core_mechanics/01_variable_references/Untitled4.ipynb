```python
a = 0
b = 2

while a < 4:
    print('a should be less that 4')
    
    a += 1
    b -= 1
    
    # Indent this entire block so it is inside the while loop
    try:
        a / b
    except ZeroDivisionError:
        print("{0}, {1} - division by 0".format(a, b))
        continue  # Now this works perfectly!
    finally:
        print("{0}, {1} - always executes".format(a, b))
```
