```python
import gc

class Node:
    def __init__(self, memory_one, memory_two=None):
        self.memory_one = memory_one
        self.memory_two = None

A = Node("Primary Server")
B = Node("Secondary Server")

A.memory_two = B # holding memory of B
B.memory_two = A # holding memory of A

del A

zombie_destroyed = gc.collect()
print (f"{zombie_destroyed}")

print("Is our Primary Server object still alive?", B.memory_two.memory_one)


        
```

```python
print("Current GC room count:", gc.get_count())
```

```python
del B

zombie_destroyed = gc.collect()
print (f"{zombie_destroyed}")

print("Current GC room count:", gc.get_count())
```
