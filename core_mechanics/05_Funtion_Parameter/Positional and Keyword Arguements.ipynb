```python
## By assigning a value directly inside the function's blueprint signature using an equals sign (parameter = default_value), you make that slot optional. If the caller provides data, Python uses it; if they don't, the function gracefully falls back to your preset system default!
```

```python
def make_meal(Burger = "McChicken", Fries = "Medium fries", Drink = "Sprite"):
    print(f"make_meal:{Burger},{Fries},{Drink}")
make_meal()
```

```python
make_meal(Burger = "McCheese")
```

```python
make_meal()
```
