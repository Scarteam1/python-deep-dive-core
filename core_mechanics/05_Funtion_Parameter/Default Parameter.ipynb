```python
from datetime import datetime

def employee_login(action, user_input=None, login_time=None):
    card_swipe = {}
    
    if login_time is None:
        login_time = datetime.now()
        
    if user_input == "swipe":
     
        card_swipe = login_time.strftime("%H:%M:%S")
    else:
        card_swipe = "No swipe recorded"
        

    print("Action Logged:", action, "| Logged In at:", card_swipe)
    
    return card_swipe


employee_name = input("Enter Employee Name (e.g. User 1, User 2): ")
action_input = input("Type 'swipe' to log in: ")

user_card = employee_login(f"{employee_name} logged in", user_input=action_input)
```
