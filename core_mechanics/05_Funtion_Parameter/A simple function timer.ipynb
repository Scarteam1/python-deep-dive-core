```python
def time(fn, *args, **kwargs):
    fn(*args, **kwargs)
```

```python
time(print, 1, 2, 3, 'Python', 'ML', sep=' --- ')
```

```python
def time(fn, *args, rep=1, **kwargs):
    for i in range(rep):
        fn(*args, **kwargs)

time(print, 1, 2, 3, 'Python', 'ML', sep=' --- ', end = ' ***\n', rep=5)
```
