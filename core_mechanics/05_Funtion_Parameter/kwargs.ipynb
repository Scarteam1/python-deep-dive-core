```python
def Query_Execution(*Query, **Connection_Status):
    print(Query)
    print(Connection_Status)
    
    for sql in Query:
        print("Executing Query", Query)
    if "TimeOut" in Connection_Status:
        print("TimeOut Error", Connection_Status)
    elif "Unable_to_establish_connection" in Connection_Status:
        print("Unable to establish connection", Connection_Status)
    else:
        print("All queries executed successfully")
```
