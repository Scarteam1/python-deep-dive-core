```python
def decorator(fn):
    def inner():
        print("---hello---")
        return fn
    return inner

@decorator
def User(Sunhik):
    print ("logged in")

User()
```

```python
def server(text):

    def my_decorators(fn):
        def inner():
            print(f"---{text}---")
            return fn
        return inner
    return my_decorators

@server("Warning")
def server_down():
    print("server is down")

@server("INFO")
def server_is_ok():
    print("server is healthy")

server_down()
server_is_ok()
```

```python
def multiply_by(factors):

    def decorators(fn):
        def inner(*args):
            answer = fn(*args)
            return answer * factors
        return inner
    return decorators

@multiply_by(10)
def add(a,b):
    return a + b

@multiply_by(100)
def substract(a,b):
    return a - b
    
print(add(10,20))
print(substract(20,10))
            
            
```
