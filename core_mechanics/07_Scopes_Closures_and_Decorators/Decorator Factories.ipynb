```python
from time import perf_counter
from functools import wraps

# Layer 1: Outer box catches your custom repetition count
def timed(reps):
    
    # Layer 2: Actual decorator catches your target function
    def dec(fn):
        
        # Layer 3: Inner wrapper runs the loop and calculates average speed
        @wraps(fn)
        def inner(*args, **kwargs):
            total_elapsed = 0
            result = None
            
            for _ in range(reps):
                start = perf_counter()
                result = fn(*args, **kwargs)
                end = perf_counter()
                total_elapsed += (end - start)
                
            avg_time = total_elapsed / reps
            print(f"{fn.__name__} took an average of {avg_time:.6f}s over {reps} runs")
            return result  # Never forget the return statement!
            
        return inner
        
    return dec

# --- Execution ---
@timed(reps=10)
def heavy_loop(n):
    total = 0
    for i in range(n):
        total += i
    return total

answer = heavy_loop(500_000)
print(f"Final Result: {answer}")
```
