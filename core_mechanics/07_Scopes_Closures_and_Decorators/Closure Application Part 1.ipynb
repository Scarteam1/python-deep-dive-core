```python
def keep_track():
    notepad = []

    def guard(new_number):
        notepad.append(new_number)
        print(notepad)
    return guard

tracker = keep_track()

tracker(10)
tracker(10)
```

```python
def white_board():
    white_board = 0

    def guard(new_number):
        nonlocal white_board
        white_board = white_board + new_number
        print(white_board)
    return guard

output = white_board()

output(10)
output(10)
```

```python
class Average:
    def __init__(self):
        self.numbers = []

    def add(self, number):
        self.numbers.append(number)
        total = sum(self.numbers)
        count = len(self.numbers)
        return total/count
```

```python
class Average:
    def __init__(self):
        self.total = 0
        self.count = 0

    def add(self, number):
        self.total += number
        self.count += 1
        return self.total/self.count
```

```python
a = Average()
```

```python
a.add(10)
```

```python
a.add(20)
```

```python
def average():
    total = 0
    count = 0
    def add(number):
        nonlocal total, count
        total = total + number
        count = count + 1
        return total/count
    return add
```

```python
a = average()
```

```python
print(a(10))
```
