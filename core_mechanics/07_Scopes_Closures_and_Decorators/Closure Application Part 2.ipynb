```python
def counter(initial_value=0):
    def inc(increment=1):
        nonlocal initial_value
        initial_value += increment
        return initial_value
    return inc
```

```python
counter1 = counter()
```

```python
counter1()
```

```python
counter1()
```

```python
def counter(fn):
    cnt = 0
    def inner(*args, **kwargs):
        nonlocal cnt
        cnt += 1
        output = fn(*args, **kwargs)
        return output
    return inner
```

```python
def add (a , b):
    return a + b

def mult (a , b):
    return a * b
```

```python
counter_add = counter(add)
counter_mult = counter(mult)
```

```python
counter_add(10, 20)
```
