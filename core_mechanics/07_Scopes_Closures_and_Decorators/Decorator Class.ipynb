```python
class CountCall():
    def __init__(self, fn):
        self.fn = fn
        self.count = 0

    def __call__(self, *args):
        self.count += 1
        print(self.count)
        return self.fn(*args)

@CountCall
def add(a, b):
    return a + b

print(add(10,2))
```

```python
class CountCall():
    def __init__(self, fn, count):
        self.fn = fn
        self.count = 0

    def __call__(self, count, *args):
        self.count += 1
        print(self.count)
        return self.fn(*args)

@CountCall
def add(a, b):
    return a + b

print(add(10,2))
```
