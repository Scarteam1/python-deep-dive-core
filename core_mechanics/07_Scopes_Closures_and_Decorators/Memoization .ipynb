```python
from functools import wraps

def memoize(fn):
    cache = {}

    @wraps(fn)
    def inner(*args):
        if args in cache:
            print(f"Fetching cache {args}")
            return cache[args]

        print(f"calculating math {args}")
        result = fn(*args)
        cache[args] = result
        return result
    return inner

@memoize
def Calculation(a, b):
    return a + b

print(Calculation(20,20))
print(Calculation(20,20))
```

```python
from functools import lru_cache

@lru_cache(maxsize = 2)

def add(a, b):
    print(f"addition of {a} and {b} is")
    return a + b


print(add(100,20))
print(add(10,20))
print(add(20,20))
```

```python
from functools import lru_cache

@lru_cache(maxsize = 100)
def fib(n):
    print('Calculating fib({0})'.format(n))
    return 1 if n < 2 else fib(n - 1) + fib(n + 1)
print(40)
print(101)
```

```python
def memoize_fib(fib):
    cache = {1:1 , 2:1}

    def inner(n):
        if n is not cache:
            cache[n] = fin[n]
        return cache[n]

@memoize_fib
def fib(n):
    print('Calculating fib({0})'.format(n))
    return 1 if n < 3 else fib(n - 1) + fib(n + 1)
print(40)
print(101)
```
