```python
from functools import wraps

def Timer1(fn):
    def inner(*args, **kwargs):
        result = fn(*args, **kwargs)
    return inner

def Timer2(fn):
    @wraps(fn)
    def inner(*args, **kwargs):
        result = fn(*args, **kwargs)
    return inner
```

```python
@Timer1
def test1():
    """This is test 1"""
    pass

@Timer2
def test2():
    """This is test 2"""
    pass

print(test1.__name__)
print(test2.__name__)
```

```python
from time import perf_counter
from functools import wraps

def Good_Timer(fn):
    @wraps(fn)
    def inner(*args, **kwargs):
        start = perf_counter()
        result = fn(*args, **kwargs)
        stop = perf_counter()

        elapsed = stop - start
        print(f"{fn.__name__} ran in {elapsed:.6f}seconds")
        return result
    return inner

@Good_Timer

def time(n):
    total = 0
    for i in range(n):
        total += 1
    return total

answer = time(100)
print(answer)
    
```
