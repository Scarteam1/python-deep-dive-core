```python
def Outer():
    outer = "Python"
    def Inner():
        nonlocal outer
        print(outer)
    return Inner
    

my_closure = Outer()

my_closure()
```

```python
def outer():
    x = 0
    def inner():
        nonlocal x
        x += 1
        print(x)
    inner()

outer() # Output: 1
outer() # Output: 1 (Reset!)
outer() # Output: 1 (Reset!)
```

```python
def outer():
    x = 0
    def inner():
        nonlocal x
        x += 1
        print(x)
    return inner

counter = outer() # outer() runs ONCE here to set x = 0

counter() # Output: 1
counter() # Output: 2 (Kept the changes!)
counter() # Output: 3
```

```python
def Outer():
    outer = "Python"
    print(hex(id(outer)))
    def Inner():
        nonlocal outer
        print(hex(id(outer)))
        print(outer)
    return Inner
    

my_closure = Outer()

my_closure()
```
