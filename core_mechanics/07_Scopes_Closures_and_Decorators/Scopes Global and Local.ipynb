```python
x = "Global Scopes"

def Scope():
    y = "Local Scope"
    print(x)
    print(y)

Scope()
```

```python
x = "Global Scopes"

def Scope():
    y = "Local Scope"
    print(x)
    print(y)

Scope()

print("x status:", "x" in globals())  # Prints: x status: True (Global remains alive)
try:
    print(y)
except NameError:
    print("y status: Dead")            # Prints: y status: Dead (Local frame was wiped from RAM)
```

```python
x = 100

def Scope():
    y = 200
    print("Local:", "y" in locals())
    print("Global", "x" in globals())
    print("Local:", "x" in locals())
    print("Global", "y" in globals())

Scope()
```

```python
status = "OK"

def emergency_shutdown():
    global status
    status = "CRITICAL"

emergency_shutdown()
print(status)
```
