```python
nums = [1, 2, 3, 4]
# Screening function: keep only even numbers
evens = filter(lambda x: x % 2 == 0, nums)

for output in evens:
    print(output)

print(evens)         # <filter object at 0x...> (Lazy placeholder)
```

```python
nums = [1, 2, 3, 4]
# Screening function: keep only even numbers
evens = filter(lambda x: x * 2 == 6, nums)

print(evens)         # <filter object at 0x...> (Lazy placeholder)
print(next(evens))   # 2 (Evaluates on demand)
```
