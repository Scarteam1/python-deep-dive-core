```python
def api_gateway( endpoint: str, network_status: str, timeout_seconds:int = 30, retry_seconds:int = 5):
    """Clean configuration telementry"""

    if network_status == "Timeout":
        print(f"Network status not stable...Retrying{retry_seconds} in seconds...")
    elif network_status == "Stable!!":
        print("Network Status stable", network_status)

    return{
        "target_url": f"www.google.com{endpoint}",
        "secure_network": True
    }

print("-----connecting server-----\n")
print (api_gateway.__doc__)
print (api_gateway.__annotations__)
print("---------------------------------\n")

print("Running Live Pipeline Transaction Output:")
print(api_gateway("/search", "Stable!!"))
```
