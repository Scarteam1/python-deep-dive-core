```python
import functools

def multiply(x, y):
    return x * y

double = functools.partial(multiply, 2)

print(double(5))
print(double(10))
```

```python
def addition(a,b):
    return a + b

add = functools.partial(addition)
print(add(3,2))
print(add(5,3))
```

```python
##Think of this as configuring a smart kitchen appliance: you set the default temperature to 350°F once, but when a unique recipe demands it, 
## you can manually type in 400°F to override that specific run without wiping out the machine's default setting.
```

```python
import functools

def Microweb(recipe, temperature = 350 ):
    return f"recipe:{recipe} Temperature{temperature'''}°F"

Oven = functools.partial(Microweb)

print(Oven("Rice", 350))
print(Oven("Dal", 400))
```
