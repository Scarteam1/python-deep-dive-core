```python
from functools import reduce
```

```python
nums = [1,2,3]
output = reduce(lambda a,b: a + b, nums)
print(output)
```

```python
nums = [1,2,3,4,5]

output = reduce(lambda a,b : a if a>b else b,nums)
print(output)
```

```python
word = 'python'

output = reduce(lambda a,b : a if a<b else b,word)
print(output)
```

```python
from functools import reduce
data = ['Python', 'is', 'Awasom']

output = reduce(lambda a, b: a + ' ' + b, data)

print(output)
```
