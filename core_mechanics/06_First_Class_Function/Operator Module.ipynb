```python
import operator

print(operator.add(2,3))
print(operator.sub(5,6))
print(operator.mul(7,8))
print(operator.mod(4,2))
print(operator.truediv(4,2))
print(operator.floordiv(4,2))
```

```python
from operator import itemgetter

items = ('abc','def','ghi')

sort = itemgetter(1)
print(sort(items))
```

```python
from operator import attrgetter

class ErrorType:
    def __init__(self, Error_Name, Error_Frequency):
        self.Error_Name = Error_Name
        self.Error_Frequency = Error_Frequency

E1 = ErrorType("Null Pointer Exception", 0.08)
E2 = ErrorType("Primary Key Constraints", 0.05)

Error_list = [E1,E2]

get_error = attrgetter("Error_Frequency")

for error_object in Error_list:
    print(get_error(error_object))
```

```python
from operator import attrgetter

class ServerStatus:
    def __init__(self, Server_Name, Server_Status):
        self.Server_Name = Server_Name
        self.Server_Status = Server_Status

S1 = ServerStatus("Server A", "True")
S2 = ServerStatus("Server B", "False")

Server_Result = attrgetter("Server_Status")

print(Server_Result(S1))
print(Server_Result(S2))
```
