```python
word = ['Python', 'ML', 'lambda']

sort_word = sorted(word, key=lambda word:len(word))
print(sort_word)
```

```python
number = [1,2,3]

sort_number = sorted(number, key=lambda number: -number)
print(sort_number)
```
