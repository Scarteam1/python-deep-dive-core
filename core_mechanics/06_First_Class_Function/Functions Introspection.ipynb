```python
import inspect
def check_status(a, b=1, c=2, *args, **kwargs) -> str:
    """The server is working fine"""
    secret_key = "Fine"
    return True

print("Shows docstring -->",check_status.__doc__)
print("This is showing what is the temporary memory address and which location the check_ststus is saved -->",check_status.__code__)
print("Local variable name -->",check_status.__name__)
print("It will show the parameters first and then the local variables as you can see -->",check_status.__code__.co_varnames)
print("This shows the number of parameters mentioned in the variable (a, b=1, c=2,) without counting the *args and **kwargs-->",check_status.__code__.co_argcount)
print(check_status.__annotations__)
print(check_status.__defaults__)
print(inspect.signature(check_status).parameters)
```
